```python
import os 
import pandas as pd

os.getcwd()
```

```python
# Load both datasets
hq = pd.read_csv("hq/high_quality_manga_sales.csv")
pop = pd.read_csv("popular/popular_manga_sales.csv")

# Union them
combined = pd.concat([hq, pop], ignore_index=True)

# Check duplicates
print(combined.duplicated(subset=["manga_id"]).sum())

combined = combined.sort_values(by="sales_circulation", na_position="last")

# Drop duplicates
combined = combined.drop_duplicates(subset=["manga_id"],keep="first")
```

```python
print(len(hq))
print(len(pop))
```

```python
# Change wrong sales data
combined.loc[combined["manga_id"] == 30013, "sales_circulation"] = 516000000
```

```python
# Verify changes
print(combined.loc[combined["manga_id"] == 30013])
```

```python
combined["sales_circulation"] = combined["sales_circulation"].str.replace(",", "")
combined["sales_circulation"] = combined["sales_circulation"].astype(float)
print(combined["sales_circulation"].head())
```

```python
# Save the unified file
print(combined.head(5))
combined.to_csv("combined_data_sales.csv", index=False)
```
